```python
duplicate_keys = [
    "Destination",
    "Date",
    "Search_Term"
]

duplicate_data = df_trends[
    df_trends.duplicated(
        subset=duplicate_keys,
        keep=False
    )
].copy()

unresolved_terms = [
    "hotels in athirappilly",
    "malampuzha garden",
    "malampuzha ropeway",
    "hotels in malampuzha",
    "konni elephant training centre"
]

unresolved_data = duplicate_data[
    duplicate_data["Search_Term"].isin(unresolved_terms)
].copy()

unresolved_data.to_csv(
    "../02_Raw_Data/google_trends_unresolved_records.csv",
    index=False
)

print("Unresolved records saved successfully!")
print("Number of records:", len(unresolved_data))
```

```python
excluded_terms = [
    "hotels in athirappilly",
    "malampuzha garden",
    "malampuzha ropeway",
    "hotels in malampuzha",
    "konni elephant training centre"
]

df_trends_clean = df_trends[
    ~df_trends["Search_Term"].isin(excluded_terms)
].copy()

print("Original records:", len(df_trends))
print("Working records:", len(df_trends_clean))
print("Excluded records:", len(df_trends) - len(df_trends_clean))

print(
    "\nRemaining duplicate combinations:",
    df_trends_clean.duplicated(
        subset=["Destination", "Date", "Search_Term"]
    ).sum()
)
```

```python
df_trends_clean.to_csv(
    "../03_Cleaned_Data/google_trends_clean.csv",
    index=False
)

print("Cleaned Google Trends dataset saved successfully!")
print("Final shape:", df_trends_clean.shape)
```

```python
destination_summary = (
    df_trends_clean.groupby("Destination")
    .agg(
        Total_Records=("Search_Term", "count"),
        Unique_Search_Terms=("Search_Term", "nunique"),
        Unique_Months=("Date", "nunique")
    )
    .reset_index()
)

print(destination_summary.to_string(index=False))

print("\nTotal destinations:",
      df_trends_clean["Destination"].nunique())

print("\nDate range:",
      df_trends_clean["Date"].min(),
      "to",
      df_trends_clean["Date"].max())
```

```python
coverage = (
    df_trends_clean.groupby(
        ["Destination", "Search_Term"]
    )
    .agg(
        Total_Records=("Date", "count"),
        Unique_Months=("Date", "nunique")
    )
    .reset_index()
)

print("Search terms with incomplete monthly coverage:")

print(
    coverage[
        coverage["Unique_Months"] < 61
    ].to_string(index=False)
)

print("\nCoverage summary:")

print(
    coverage["Unique_Months"].value_counts().sort_index()
)
```

```python
category_summary = (
    df_trends_clean.groupby(
        ["Destination", "Category"]
    )
    .agg(
        Average_Score=("Google_Trends_Score", "mean"),
        Maximum_Score=("Google_Trends_Score", "max"),
        Total_Search_Terms=("Search_Term", "nunique")
    )
    .reset_index()
)

category_summary["Average_Score"] = (
    category_summary["Average_Score"].round(2)
)

print(category_summary.head(20).to_string(index=False))
```

```python
normalization_summary = (
    df_trends_clean.groupby(
        ["Category", "Normalization_Group"]
    )
    .agg(
        Total_Records=("Google_Trends_Score", "count"),
        Unique_Destinations=("Destination", "nunique"),
        Unique_Search_Terms=("Search_Term", "nunique")
    )
    .reset_index()
)

print(
    normalization_summary.to_string(index=False)
)
```

```python
valid_groups = [
    "accommodation",
    "food",
    "activities",
    "destination_interest"
]

df_valid_trends = df_trends_clean[
    df_trends_clean["Normalization_Group"].isin(valid_groups)
].copy()

print("Records in named normalization groups:",
      len(df_valid_trends))

print("\nDestination coverage:")

print(
    df_valid_trends.groupby(
        "Normalization_Group"
    )["Destination"].nunique()
)

print("\nSearch term coverage:")

print(
    df_valid_trends.groupby(
        "Normalization_Group"
    )["Search_Term"].nunique()
)

print("\nDestination and group coverage:")

print(
    df_valid_trends.groupby(
        ["Destination", "Normalization_Group"]
    )["Search_Term"].nunique().to_string()
)
```

```python
group_summary = (
    df_valid_trends.groupby(
        ["Normalization_Group", "Date"]
    )
    .agg(
        Average_Score=("Google_Trends_Score", "mean"),
        Minimum_Score=("Google_Trends_Score", "min"),
        Maximum_Score=("Google_Trends_Score", "max"),
        Search_Terms=("Search_Term", "nunique")
    )
    .reset_index()
)

print(group_summary.head(20).to_string(index=False))

print("\nNumber of months per normalization group:")

print(
    group_summary.groupby("Normalization_Group")["Date"].nunique()
)
```

```python
metadata_summary = (
    df_valid_trends.groupby(
        [
            "Normalization_Group",
            "Selection_Type",
            "Source_File"
        ]
    )
    .agg(
        Records=("Google_Trends_Score", "count"),
        Destinations=("Destination", "nunique"),
        Search_Terms=("Search_Term", "nunique")
    )
    .reset_index()
)

print(metadata_summary.to_string(index=False))
```

```python
for destination in ["Munnar", "Thekkady", "Vagamon", "Wayanad"]:

    print("\nDESTINATION:", destination)

    data = df_valid_trends[
        df_valid_trends["Destination"] == destination
    ]

    print(
        data.groupby(
            ["Normalization_Group", "Selection_Type"]
        )["Search_Term"]
        .unique()
        .to_string()
    )
```

```python
group_coverage = (
    df_trends_clean.groupby(
        ["Destination", "Normalization_Group"]
    )
    .agg(
        Records=("Google_Trends_Score", "count"),
        Search_Terms=("Search_Term", "nunique")
    )
    .reset_index()
)

coverage_table = group_coverage.pivot(
    index="Destination",
    columns="Normalization_Group",
    values="Search_Terms"
).fillna(0)

print(coverage_table.to_string())
```

```python
category_summary.to_csv(
    "../06_EDA_Analysis/google_trends_category_summary.csv",
    index=False
)

coverage_table.to_csv(
    "../06_EDA_Analysis/google_trends_normalization_coverage.csv"
)

print("Google Trends analysis summaries saved successfully!")
```
